```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. KONFIGURASI VISUALISASI
sns.set_theme(style="white", rc={"axes.facecolor": "#ffffff"})
plt.figure(figsize=(10, 6))

# 2. MEMBACA DATASET
df = pd.read_csv('Kelas A_Gender Inequality Index.csv')

# 3. PEMBERSIHAN DAN AGREGASI DATA
# Menghapus missing values pada kolom yang akan dianalisis
df_clean = df.dropna(subset=['Human_development', 'Maternal_mortality'])

# Menghitung nilai maksimum dan mengurutkan dari tertinggi ke terendah
max_data = df_clean.groupby('Human_development')['Maternal_mortality'].max().reset_index()
max_data = max_data.sort_values('Maternal_mortality', ascending=False)

# 4. PEMBUATAN VISUALISASI
# Membuat bar chart dengan palet warna 'rocket'
ax = sns.barplot(
    x='Human_development', 
    y='Maternal_mortality', 
    data=max_data, 
    palette='rocket', 
    edgecolor='black',
    linewidth=1.2,
    width=0.6 
)

# 5. KUSTOMISASI ESTETIKA
# Menghilangkan garis tepi atas dan kanan
sns.despine()
# Menambahkan garis bantu grid pada sumbu Y
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Mengatur judul dan label sumbu
plt.title('Nilai Maksimum Maternal Mortality\nBerdasarkan Tingkat Human Development', 
          fontsize=15, fontweight='bold', pad=20)
plt.xlabel('Tingkat Human Development', fontsize=12, fontweight='bold', labelpad=10)
plt.ylabel('Maksimum Maternal Mortality', fontsize=12, fontweight='bold', labelpad=10)

# Menambahkan anotasi nilai (angka maksimum) tepat di atas setiap bar
for p in ax.patches:
    height = p.get_height()
    ax.annotate(f'{int(height)}', 
                (p.get_x() + p.get_width() / 2., height), 
                ha='center', va='bottom', 
                xytext=(0, 8), textcoords='offset points',
                fontsize=11, fontweight='bold')

plt.tight_layout()

# 6. PENYIMPANAN GAMBAR
plt.savefig('Soal_1_Maksimum_Maternal_Mortality.png', dpi=300, bbox_inches='tight')
print("Selesai! File 'Soal_1_Maksimum_Maternal_Mortality.png' berhasil disimpan.")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. KONFIGURASI VISUALISASI
sns.set_theme(style="whitegrid", rc={"axes.facecolor": "#ffffff"})
# Kanvas dibikin agak lebar karena jumlah negaranya mungkin lumayan banyak
plt.figure(figsize=(15, 7))

# 2. MEMBACA DATASET
df = pd.read_csv('Kelas A_Gender Inequality Index.csv')

# 3. PEMBERSIHAN DAN FILTERING DATA
# Menghapus missing values pada kolom yang akan dianalisis
df_clean = df.dropna(subset=['Country', 'Seats_parliament', 'GII'])

# Memfilter negara dengan keterwakilan perempuan di parlemen > 40%
df_filtered = df_clean[df_clean['Seats_parliament'] > 40]

# Mengurutkan berdasarkan GII terbaik (terendah ke tertinggi)
# Semakin rendah nilai GII, semakin baik tingkat kesetaraan gendernya
df_filtered = df_filtered.sort_values(by='GII', ascending=True)

# 4. PEMBUATAN VISUALISASI
# Menggunakan bar chart untuk menampilkan GII dari negara terpilih
ax = sns.barplot(
    x='Country', 
    y='GII', 
    data=df_filtered, 
    palette='viridis', 
    edgecolor='black',
    linewidth=1.2
)

# 5. KUSTOMISASI ESTETIKA
sns.despine()

# Mengatur judul dan label sumbu
plt.title('Negara dengan Keterwakilan Parlemen Perempuan > 40%\nDiurutkan dari Nilai GII Terbaik', 
          fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Negara', fontsize=12, fontweight='bold', labelpad=10)
plt.ylabel('Gender Inequality Index (GII)', fontsize=12, fontweight='bold', labelpad=10)

# Memutar label nama negara agar tidak bertabrakan
plt.xticks(rotation=45, ha='right', fontsize=10)

# Menambahkan anotasi nilai GII pada setiap bar
# Menggunakan 3 angka di belakang koma (.3f) karena rentang GII sangat kecil
for p in ax.patches:
    height = p.get_height()
    ax.annotate(f'{height:.3f}', 
                (p.get_x() + p.get_width() / 2., height), 
                ha='center', va='bottom', 
                xytext=(0, 5), textcoords='offset points',
                fontsize=9, fontweight='bold')

plt.tight_layout()

# 6. PENYIMPANAN GAMBAR
plt.savefig('Soal_2_Filter_Kursi_Parlemen_GII.png', dpi=300, bbox_inches='tight')
print("Selesai! File 'Soal_2_Filter_Kursi_Parlemen_GII.png' berhasil disimpan.")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. KONFIGURASI VISUALISASI
sns.set_theme(style="whitegrid", rc={"axes.facecolor": "#ffffff"})
plt.figure(figsize=(10, 6))

# 2. MEMBACA DATASET
df = pd.read_csv('Kelas A_Gender Inequality Index.csv')

# 3. PEMBERSIHAN DATA
# Menghapus missing values pada kolom yang akan dianalisis korelasinya
df_clean = df.dropna(subset=['F_Labour_force', 'GII'])

# 4. PEMBUATAN VISUALISASI (SCATTER PLOT + REGPLOT)
# Membuat garis tren (regresi)
# Parameter scatter=False agar titik bawaan regplot tidak ikut tergambar
sns.regplot(
    x='F_Labour_force', 
    y='GII', 
    data=df_clean, 
    scatter=False, 
    color='red', 
    line_kws={'linestyle':'--', 'linewidth': 2}
)

# Membuat scatter plot untuk titik data
sns.scatterplot(
    x='F_Labour_force', 
    y='GII', 
    data=df_clean, 
    color='#2a9d8f', # Warna hijau toska yang clean
    s=80, # Ukuran titik
    alpha=0.7, # Transparansi agar titik yang bertumpuk tetap terlihat
    edgecolor='black'
)

# 5. KUSTOMISASI ESTETIKA
sns.despine()

# Mengatur judul dan label sumbu
plt.title('Korelasi Angkatan Kerja Perempuan (F_Labour_force)\ndengan Gender Inequality Index (GII)', 
          fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Angkatan Kerja Perempuan (%)', fontsize=12, fontweight='bold', labelpad=10)
plt.ylabel('Gender Inequality Index (GII)', fontsize=12, fontweight='bold', labelpad=10)

plt.tight_layout()

# 6. PENYIMPANAN GAMBAR
plt.savefig('Soal_3_Korelasi_F_Labour_GII.png', dpi=300, bbox_inches='tight')
print("Selesai! File 'Soal_3_Korelasi_F_Labour_GII.png' berhasil disimpan.")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. KONFIGURASI VISUALISASI
sns.set_theme(style="whitegrid", rc={"axes.facecolor": "#ffffff"})
# Kanvas dibuat memanjang secara horizontal agar boxplot lebih jelas dibaca
plt.figure(figsize=(12, 5))

# 2. MEMBACA DATASET
df = pd.read_csv('Kelas A_Gender Inequality Index.csv')

# 3. PEMBERSIHAN DATA
# Menghapus missing values pada kolom keterwakilan parlemen
df_clean = df.dropna(subset=['Seats_parliament'])

# 4. PEMBUATAN VISUALISASI (BOXPLOT)
# Membuat boxplot dengan penyesuaian gaya khusus untuk mendeteksi outlier
sns.boxplot(
    x=df_clean['Seats_parliament'], 
    color='#457b9d', # Warna biru kalem untuk kotak utama
    width=0.4,
    fliersize=8, # Memperbesar ukuran titik pencilan (outlier)
    flierprops={'marker': 'o', 'markerfacecolor': 'red', 'markeredgecolor': 'black'} # Pencilan diberi warna merah
)

# 5. KUSTOMISASI ESTETIKA
sns.despine(left=True) # Menghilangkan garis tepi kiri agar lebih bersih

# Mengatur judul dan label sumbu
plt.title('Sebaran Data Keterwakilan Perempuan di Parlemen\n(Deteksi Negara Pencilan / Outlier)', 
          fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Persentase Kursi Parlemen (%)', fontsize=12, fontweight='bold', labelpad=10)

plt.tight_layout()

# 6. PENYIMPANAN GAMBAR
plt.savefig('Soal_4_Boxplot_Pencilan_Parlemen.png', dpi=300, bbox_inches='tight')
print("Selesai! File 'Soal_4_Boxplot_Pencilan_Parlemen.png' berhasil disimpan.")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. KONFIGURASI FIGUR UTAMA (GRID 2x2)
sns.set_theme(style="white", rc={"axes.facecolor": "#ffffff"})
# Ukuran kanvas dibuat besar agar 4 grafik tidak saling bertumpuk
fig, axes = plt.subplots(2, 2, figsize=(22, 14))

# Menambahkan judul utama dashboard
fig.suptitle('Dashboard Analisis Gender Inequality Index (GII)', fontsize=24, fontweight='bold', y=1.02)

# 2. MEMBACA DATASET
df = pd.read_csv('Kelas A_Gender Inequality Index.csv')

# =====================================================================
# GRAFIK 1: KIRI ATAS (axes[0, 0]) - Maksimum Maternal Mortality
# =====================================================================
df_1 = df.dropna(subset=['Human_development', 'Maternal_mortality'])
max_data = df_1.groupby('Human_development')['Maternal_mortality'].max().reset_index()
max_data = max_data.sort_values('Maternal_mortality', ascending=False)

sns.barplot(x='Human_development', y='Maternal_mortality', data=max_data, 
            palette='rocket', edgecolor='black', linewidth=1.2, width=0.6, ax=axes[0, 0])

axes[0, 0].set_title('Nilai Maksimum Maternal Mortality', fontsize=16, fontweight='bold', pad=15)
axes[0, 0].set_xlabel('Tingkat Human Development', fontweight='bold')
axes[0, 0].set_ylabel('Maksimum Maternal Mortality', fontweight='bold')
axes[0, 0].grid(axis='y', linestyle='--', alpha=0.5)

# Anotasi nilai maksimum pada bar
for p in axes[0, 0].patches:
    axes[0, 0].annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()), 
                        ha='center', va='bottom', xytext=(0, 8), textcoords='offset points', fontweight='bold')

# =====================================================================
# GRAFIK 2: KANAN ATAS (axes[0, 1]) - Keterwakilan Parlemen > 40% & GII Terbaik
# =====================================================================
df_2 = df.dropna(subset=['Country', 'Seats_parliament', 'GII'])
df_filtered = df_2[df_2['Seats_parliament'] > 40].copy()
df_filtered = df_filtered.sort_values(by='GII', ascending=True)

sns.barplot(x='Country', y='GII', data=df_filtered, 
            palette='viridis', edgecolor='black', linewidth=1.2, ax=axes[0, 1])

axes[0, 1].set_title('Negara Keterwakilan Parlemen > 40% (GII Terbaik)', fontsize=16, fontweight='bold', pad=15)
axes[0, 1].set_xlabel('Negara', fontweight='bold')
axes[0, 1].set_ylabel('Gender Inequality Index (GII)', fontweight='bold')
axes[0, 1].grid(axis='y', linestyle='--', alpha=0.5)
axes[0, 1].tick_params(axis='x', labelrotation=45) # Rotasi nama negara agar terbaca

# Anotasi nilai GII (3 angka di belakang koma)
for p in axes[0, 1].patches:
    axes[0, 1].annotate(f'{p.get_height():.3f}', (p.get_x() + p.get_width() / 2., p.get_height()), 
                        ha='center', va='bottom', xytext=(0, 5), textcoords='offset points', fontsize=9, fontweight='bold')

# =====================================================================
# GRAFIK 3: KIRI BAWAH (axes[1, 0]) - Korelasi Angkatan Kerja vs GII
# =====================================================================
df_3 = df.dropna(subset=['F_Labour_force', 'GII']).copy()

# Regresi (Garis Tren) & Scatter Plot
sns.regplot(x='F_Labour_force', y='GII', data=df_3, scatter=False, color='red', 
            line_kws={'linestyle':'--', 'linewidth': 2}, ax=axes[1, 0])
sns.scatterplot(x='F_Labour_force', y='GII', data=df_3, color='#2a9d8f', 
                s=80, alpha=0.7, edgecolor='black', ax=axes[1, 0])

axes[1, 0].set_title('Korelasi Angkatan Kerja Perempuan vs GII', fontsize=16, fontweight='bold', pad=15)
axes[1, 0].set_xlabel('Angkatan Kerja Perempuan (%)', fontweight='bold')
axes[1, 0].set_ylabel('Gender Inequality Index (GII)', fontweight='bold')
axes[1, 0].grid(True, linestyle='--', alpha=0.5)

# =====================================================================
# GRAFIK 4: KANAN BAWAH (axes[1, 1]) - Boxplot Deteksi Pencilan Parlemen
# =====================================================================
df_4 = df.dropna(subset=['Seats_parliament']).copy()

sns.boxplot(x=df_4['Seats_parliament'], color='#457b9d', width=0.4, 
            fliersize=8, flierprops={'marker': 'o', 'markerfacecolor': 'red', 'markeredgecolor': 'black'}, ax=axes[1, 1])

axes[1, 1].set_title('Sebaran Data Keterwakilan Perempuan di Parlemen (Outlier)', fontsize=16, fontweight='bold', pad=15)
axes[1, 1].set_xlabel('Persentase Kursi Parlemen (%)', fontweight='bold')
axes[1, 1].grid(axis='x', linestyle='--', alpha=0.5)

# =====================================================================
# FINALISASI & PENYIMPANAN
# =====================================================================
sns.despine()
plt.tight_layout()

# Menyimpan hasil dashboard ke format PNG
plt.savefig('Soal_5_Dashboard_Gabungan_Kelompok_B.png', dpi=300, bbox_inches='tight')

print("Proses selesai. File 'Soal_5_Dashboard_Gabungan_Kelompok_B.png' berhasil disimpan.")
```
